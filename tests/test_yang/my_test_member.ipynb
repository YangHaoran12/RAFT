```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def draw_cylinder(z, radius):
    # Create a grid for the cylinder
    theta = np.linspace(0, 2 * np.pi, 100)
    theta_grid, z_grid = np.meshgrid(theta, z)
    
    # Create the cylinder's x and y coordinates based on the radius
    x_grid = radius[:, np.newaxis] * np.cos(theta_grid)
    y_grid = radius[:, np.newaxis] * np.sin(theta_grid)

    return x_grid, y_grid, z_grid

# Define the height and changing radius
z = np.linspace(0, 10, 100)
radius = np.linspace(1, 3, 100)  # Change diameter from 1 to 3

# Draw the cylinder
x, y, z = draw_cylinder(z, radius)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, alpha=0.5)

# Set labels
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('Cylinder with Changing Diameter')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def draw_cylinder(z, radius):
    theta = np.linspace(0, 2 * np.pi, 100)
    theta_grid, z_grid = np.meshgrid(theta, z)
    x_grid = radius[:, np.newaxis] * np.cos(theta_grid)
    y_grid = radius[:, np.newaxis] * np.sin(theta_grid)

    return x_grid, y_grid, z_grid

# Define the height and changing radius
z = np.linspace(0, 10, 100)
radius = np.linspace(1, 3, 100)  # Change diameter from 1 to 3

# Draw the cylinder
x, y, z_cylinder = draw_cylinder(z, radius)

# Create a grid for the plane z = 0
x_plane, y_plane = np.meshgrid(np.linspace(-4, 4, 100), np.linspace(-4, 4, 100))
z_plane = np.zeros_like(x_plane)

# Set up the figure and 3D axis
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the cylinder
ax.plot_surface(x, y, z_cylinder, alpha=0.5, color='blue')

# Plot the plane z = 0
ax.plot_surface(x_plane, y_plane, z_plane, alpha=0.5, color='gray')

# Define the rectangular prism vertices
x_rect = [1, 1, -1, -1, 1]
y_rect = [1, -1, -1, 1, 1]
z_rect = [5, 5, 5, 5, 5]  # Top face at z=5

# Draw the rectangular prism faces
# Top face
ax.plot_surface(x_rect[:4], y_rect[:4], z_rect[:4], alpha=0.5, color='red')
# Bottom face
ax.plot_surface(x_rect[:4], y_rect[:4], np.zeros(4), alpha=0.5, color='orange')
# Side faces
ax.plot_surface([x_rect[0], x_rect[1], x_rect[1], x_rect[0]], 
                [y_rect[0], y_rect[0], y_rect[0], y_rect[0]], 
                [5, 5, 0, 0], alpha=0.5, color='green')
ax.plot_surface([x_rect[1], x_rect[2], x_rect[2], x_rect[1]], 
                [y_rect[1], y_rect[1], y_rect[1], y_rect[1]], 
                [5, 5, 0, 0], alpha=0.5, color='blue')
ax.plot_surface([x_rect[2], x_rect[3], x_rect[3], x_rect[2]], 
                [y_rect[2], y_rect[2], y_rect[2], y_rect[2]], 
                [5, 5, 0, 0], alpha=0.5, color='purple')
ax.plot_surface([x_rect[3], x_rect[0], x_rect[0], x_rect[3]], 
                [y_rect[3], y_rect[3], y_rect[3], y_rect[3]], 
                [5, 5, 0, 0], alpha=0.5, color='yellow')

# Set limits for each axis
ax.set_xlim([-4, 4])
ax.set_ylim([-4, 4])
ax.set_zlim([0, 10])

# Set view angle
ax.view_init(elev=30, azim=30)

# Set labels
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('Cylinder with Changing Diameter, Plane z=0, and Rectangular Prism')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 圆台的参数
r1 = 10  # 底面半径
r2 = 4   # 顶面半径
h = 20   # 高度

# 切割平面的位置
d = 8    # 切平面距离底面的高度

# 切平面处的半径
r_d = r1 - (r1 - r2) * (d / h)

# 生成椭圆的角度
theta = np.linspace(0, 2 * np.pi, 100)

# 椭圆的 x 和 y 坐标
x = r_d * np.cos(theta)
y = r_d * np.sin(theta)

# 绘图
plt.figure(figsize=(6, 6))
plt.plot(x, y, label=f"Elliptical Cross-Section at d={d} units", color='b')
plt.fill(x, y, 'b', alpha=0.2)

# 绘制圆台的轴线
plt.axhline(0, color='k', linestyle='--', label="Center Line (Axis)")

# 添加标题和标签
plt.title(f"Elliptical Cross-Section of a Frustum at d={d}")
plt.xlabel('X')
plt.ylabel('Y')
plt.gca().set_aspect('equal', adjustable='box')  # 使比例相等，保证椭圆不变形
plt.legend()

# 显示图形
plt.grid(True)
plt.show()
```
