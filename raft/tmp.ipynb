```python
import numpy as np
from scipy.integrate import quad

# 定义每个惯性矩的积分函数
def I_xx_integrand(z, a0, a1, b0, b1, h):
    b_z = b0 + (b1 - b0) * z / h  # b(z)
    a_z = a0 + (a1 - a0) * z / h  # a(z)
    return (1/4) * np.pi * b_z*a_z*(b_z**2+a_z**2)/2  # I_xx(z)

def I_yy_integrand(z, a0, a1, b0, b1, h):
    a_z = a0 + (a1 - a0) * z / h  # a(z)
    b_z = b0 + (b1 - b0) * z / h  # b(z)
    return (1/4) * np.pi * a_z*b_z*(b_z**2+a_z**2)/2  # I_yy(z)

def I_zz_integrand(z, a0, a1, b0, b1, h):
    a_z = a0 + (a1 - a0) * z / h  # a(z)
    b_z = b0 + (b1 - b0) * z / h  # b(z)
    a2_b2 = a_z**2 + b_z**2
    return (1/2) * np.pi * a_z * b_z * (a2_b2 / 2)  # I_zz(z)

# 定义主函数来计算总惯性矩
def calculate_inertia_moments(a0, a1, b0, b1, h):
    # 计算 I_xx
    I_xx, _ = quad(I_xx_integrand, 0, h, args=(a0, a1, b0, b1, h))
    
    # 计算 I_yy
    I_yy, _ = quad(I_yy_integrand, 0, h, args=(a0, a1, b0, b1, h))
    
    # 计算 I_zz
    I_zz, _ = quad(I_zz_integrand, 0, h, args=(a0, a1, b0, b1, h))
    
    return I_xx, I_yy, I_zz

# 示例：给定椭圆台体的参数
a0, a1 = 3, 3   # 底部和顶部的半长轴
b0, b1 = 3, 3   # 底部和顶部的半短轴
h = 12          # 台体的高度

# 计算惯性矩
I_xx, I_yy, I_zz = calculate_inertia_moments(a0, a1, b0, b1, h)

V, hc = EllipticalFrustumVCV([6, 6], [6, 6], H)

# 输出结果
print(f"I_xx: {I_xx*7850:.2f}")
print(f"I_yy: {I_yy*7850:.2f}")
print(f"I_zz: {I_zz*7850:.2f}")
```

```python
import numpy as np
from scipy.integrate import tplquad

# 定义参数
a0 = 3  # 底部的半长轴
b0 = 3  # 底部的半短轴
a1 = 3  # 顶部的半长轴
b1 = 3  # 顶部的半短轴
H = 12.0  # 圆台的高度

# 定义 r(θ) 公式
def r_theta(a0, b0, a1, b1, H, z, theta):
    # a(z) 和 b(z) 是根据 z 来变化的
    a_z = (a0+a1)/2 + (a1 - a0) * z / H  # 随着z变化的 a(z)
    b_z = (b0+b1)/2 + (b1 - b0) * z / H  # 随着z变化的 b(z)
    return (a_z * b_z) / np.sqrt(b_z**2 * np.cos(theta)**2 + a_z**2 * np.sin(theta)**2)

def r_cicular(a0, b0, a1, b1, H, z, theta):
    a_z = a0 + (a1 - a0) * z / H  # 随着z变化的 a(z)
    b_z = b0 + (b1 - b0) * z / H  # 随着z变化的 b(z)
    return a_z

# 被积函数
def integrand(r, theta, z, a0, b0, a1, b1, H):
    return (r**2 * np.sin(theta)**2 + z**2) * r

# 三重积分的积分顺序是 r, theta, z
def triple_integral(a0, b0, a1, b1, H):
    # 使用 tplquad 进行三重积分
    result, error = tplquad(
        lambda r, theta, z: integrand(r, theta, z, a0, b0, a1, b1, H),
        -H/2, H/2,  # z 从 0 到 H
        lambda z: 0,  # θ 的下限是 0
        lambda z: 2 * np.pi,  # θ 的上限是 2π
        lambda z, theta: 0,  # r 的下限是 0
        lambda z, theta: r_theta(a0, b0, a1, b1, H, z, theta)  # r 的上限是 r(θ)
    )
    return result

# 计算最终的积分值
final_result = triple_integral(a0, b0, a1, b1, H)
print(f"Triple integral result: {final_result*7850:.6f}")

from raft.helpers import EllipticalFrustumMOI, FrustumMOI, FrustumVCV, EllipticalFrustumVCV

I_xx, I_yy, I_zz = EllipticalFrustumMOI([a0*2,b0*2], [a1*2,b1*2], H, 7850)
print(f"Triple integral result: {I_xx:.6f}")
print(f"Triple integral result: {I_yy:.6f}")
print(f"Triple integral result: {I_zz:.6f}")

I_rad, I_ax = FrustumMOI(a0*2, a1*2, H, 7850)
print(f"I_rad: {I_rad:.6f}")
print(f"I_ax : {I_ax:.6f}")

I_rad -= FrustumVCV(a0*2, a1*2, H, rtn=1)*7850 * FrustumVCV(a0*2, a1*2, H, rtn=2)**2

V, hc = EllipticalFrustumVCV([a0*2,b0*2], [a1*2,b1*2], H)
print(hc)
I_xx -= V*7850 * hc**2
print(f"I_rad: {I_rad:.6f}")
print(f"I_xx : {I_xx:.6f}")
```
