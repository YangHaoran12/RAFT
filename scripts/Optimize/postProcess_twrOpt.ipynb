```python
import openmdao.api as om
from raft.raft_rotor import raft_dir
import os.path as osp

sql_dir = osp.join(raft_dir, "rotors_opt_out/tower.sql")

cr = om.CaseReader(sql_dir)


print("Available cases: ", cr.list_cases())

# 获取 driver 层面的所有案例（常用于优化迭代）
driver_cases = cr.list_cases('driver')

# 读取第一个案例的数据
case = cr.get_case(driver_cases[1])
print("Inputs: ", case.inputs)      # 输入数据
print("Outputs: ", case.outputs)    # 输出数据
```

```python
import matplotlib.pyplot as plt

# 提取目标函数值
obj_values = [cr.get_case(case_id).get_val('twin_rotor_comp.tower_mass') for case_id in driver_cases]

plt.plot(obj_values, marker='o')
plt.xlabel('Iteration')
plt.ylabel('Objective')
plt.title('Optimization Convergence')
plt.show()
```

```python
import matplotlib.pyplot as plt

# 提取目标函数值
obj_values = [cr.get_case(case_id).get_val('twin_rotor_comp.max_tower_stress') for case_id in driver_cases]

plt.plot(obj_values, marker='o')
plt.xlabel('Iteration')
plt.ylabel('Objective')
plt.title('Optimization Convergence')
plt.show()
```
